```python
import numpy as np
array=np.random.randint(1,100,16)
array
```

```python
array.mean()
```

```python
np.percentile(array,25)
```

```python
np.percentile(array,50)
```

```python
np.percentile(array,75)
```

```python
np.percentile(array,100)
```

```python
def outDetection(array):
  sorted(array)
  Q1,Q3=np.percentile(array,[25,75])
  IQR=Q3-Q1
  lr=Q1-(1.5*IQR)
  ur=Q3+(1.5*IQR)
  return lr,ur
lr,ur=outDetection(array)
lr,ur
```

```python
import seaborn as sns
%matplotlib inline
sns.displot(array)
```

```python
sns.histplot(array, kde=True)
```

```python
new_array=array[(array>lr) & (array<ur)]
new_array
```

```python
sns.displot(new_array)
```

```python
lr1,ur1=outDetection(new_array)
lr1,ur1
```

```python
final_array=new_array[(new_array>lr1) & (new_array<ur1)]
final_array
```

```python
sns.displot(final_array)
```
